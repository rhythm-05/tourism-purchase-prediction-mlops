## **Data Description**

The dataset contains customer and interaction data that serve as key attributes for predicting the likelihood of purchasing the Wellness Tourism Package. The detailed attributes are:

**Customer Details**
- **CustomerID:** Unique identifier for each customer.
- **ProdTaken:** Target variable indicating whether the customer has purchased a package (0: No, 1: Yes).
- **Age:** Age of the customer.
- **TypeofContact:** The method by which the customer was contacted (Company Invited or Self Inquiry).
- **CityTier:** The city category based on development, population, and living standards (Tier 1 > Tier 2 > Tier 3).
- **Occupation:** Customer's occupation (e.g., Salaried, Freelancer).
- **Gender:** Gender of the customer (Male, Female).
- **NumberOfPersonVisiting:** Total number of people accompanying the customer on the trip.
- **PreferredPropertyStar:** Preferred hotel rating by the customer.
- **MaritalStatus:** Marital status of the customer (Single, Married, Divorced).
- **NumberOfTrips:** Average number of trips the customer takes annually.
- **Passport:** Whether the customer holds a valid passport (0: No, 1: Yes).
- **OwnCar:** Whether the customer owns a car (0: No, 1: Yes).
- **NumberOfChildrenVisiting:** Number of children below age 5 accompanying the customer.
- **Designation:** Customer's designation in their current organization.
- **MonthlyIncome:** Gross monthly income of the customer.

**Customer Interaction Data**
- **PitchSatisfactionScore:** Score indicating the customer's satisfaction with the sales pitch.
- **ProductPitched:** The type of product pitched to the customer.
- **NumberOfFollowups:** Total number of follow-ups by the salesperson after the sales pitch.-
- **DurationOfPitch:** Duration of the sales pitch delivered to the customer.

## Data Preparation

````python
import pandas as pd
df = pd.read_csv("tourism.csv")
df.head()
````

````python
# Check the shape of the dataset
print("Dataset Shape:", df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check basic information about the dataset
print("\nDataset Information:")
df.info()
````

### Initial Data Quality Check

The dataset contains 4,128 observations and 21 columns. The data types include numerical and categorical features. No missing values are present based on the dataset information.

We will further check for duplicate records, target class distribution, and the unique values of categorical variables before performing preprocessing.

````python
# Check for missing values
print("Missing Values:")
print(df.isnull().sum())

# Check for duplicate rows
print("\nNumber of Duplicate Rows:", df.duplicated().sum())

# Check target variable distribution
print("\nTarget Variable Distribution:")
print(df["ProdTaken"].value_counts())

print("\nTarget Variable Percentage:")
print(df["ProdTaken"].value_counts(normalize=True) * 100)

# Check unique values of categorical columns
categorical_cols = df.select_dtypes(include="object").columns

print("\nUnique Values in Categorical Columns:")
for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].unique())
````

### Data Quality Findings

The dataset contains no missing values and no duplicate rows.

The target variable `ProdTaken` is imbalanced, with 80.69% of customers not purchasing the package and 19.31% purchasing it. Therefore, class imbalance should be considered while training and evaluating the classification models.

Among the categorical variables, the `Gender` column contains an inconsistent category, `Fe Male`, which should be standardized with `Female` before model training.

The `Unnamed: 0` column appears to be an index column and is not expected to provide useful predictive information. `CustomerID` is an identifier and should also be evaluated before being used as a model feature.

````python
# Check whether CustomerID is unique
print("Unique CustomerID values:", df["CustomerID"].nunique())
print("Total rows:", len(df))

# Check numerical feature statistics
print("\nNumerical Feature Summary:")
print(df.describe().T)

# Check value counts for Gender
print("\nGender Distribution:")
print(df["Gender"].value_counts())
````

### Data Cleaning Decisions

The `CustomerID` column contains a unique value for every observation and is an identifier rather than a predictive feature. The `Unnamed: 0` column is also an index-like column. Both columns will therefore be removed before model training.

The `Gender` column contains the categories `Male`, `Female`, and `Fe Male`. The `Fe Male` category is an inconsistent representation of `Female`, so it will be standardized to `Female`.

The numerical summary also shows some potentially high values in features such as `DurationOfPitch` and `MonthlyIncome`. These will be examined carefully rather than removed automatically.

````python
# Create a copy for data preparation
df_clean = df.copy()

# Remove identifier/index columns
df_clean.drop(columns=["Unnamed: 0", "CustomerID"], inplace=True)

# Standardize inconsistent Gender category
df_clean["Gender"] = df_clean["Gender"].replace({"Fe Male": "Female"})

# Check the cleaned structure
print("Shape after removing identifier columns:", df_clean.shape)

print("\nGender values after standardization:")
print(df_clean["Gender"].value_counts())

print("\nRemaining columns:")
print(df_clean.columns.tolist())
````

### Exploratory Feature Analysis

After removing the identifier columns and standardizing the inconsistent gender category, the remaining features are divided into numerical and categorical variables.

Before applying transformations, the distributions of numerical variables and the frequency of categorical variables will be examined to identify potential skewness, unusual values, or categories that may require special handling.

````python
# Separate target variable
target = "ProdTaken"

# Identify numerical and categorical columns
numerical_cols = df_clean.drop(columns=[target]).select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_cols = df_clean.drop(columns=[target]).select_dtypes(
    include=["object"]
).columns.tolist()

print("Numerical Columns:")
print(numerical_cols)

print("\nCategorical Columns:")
print(categorical_cols)

# Check skewness of numerical features
print("\nNumerical Feature Skewness:")
print(df_clean[numerical_cols].skew().sort_values(ascending=False))

# Check categorical distributions
print("\nCategorical Feature Distributions:")
for col in categorical_cols:
    print(f"\n{col}:")
    print(df_clean[col].value_counts())
````

### Preprocessing Strategy

The dataset contains 12 numerical features and 6 categorical features after removing the identifier columns.

The categorical variables will be encoded using one-hot encoding so that they can be used by the machine learning algorithms. Unknown categories during inference will be handled safely.

The numerical features will be passed through without transformation. Although some numerical variables show positive skewness, the selected models are tree-based algorithms and generally do not require feature scaling or normally distributed numerical variables.

The target variable `ProdTaken` will be separated from the input features before model training.

````python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Separate features and target
X = df_clean.drop(columns=["ProdTaken"])
y = df_clean["ProdTaken"]

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Identify categorical and numerical columns
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()
numerical_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_cols
        )
    ],
    remainder="passthrough"
)

# Fit preprocessing only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using the fitted preprocessor
X_test_processed = preprocessor.transform(X_test)

print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)

print("\nProcessed training set shape:", X_train_processed.shape)
print("Processed testing set shape:", X_test_processed.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))
````

````python
feature_names = preprocessor.get_feature_names_out()

print("Number of transformed features:", len(feature_names))

print("\nTransformed feature names:")
print(feature_names)
````

### Transformed Feature Set

The preprocessing pipeline transformed the original 18 input features into 34 model-ready features.

The six categorical variables were encoded using One-Hot Encoding, resulting in 22 binary features. The remaining 12 numerical features were retained as numerical variables.

The transformed feature set therefore contains 34 features in total. The `handle_unknown="ignore"` option was used in the encoder to ensure that unseen categorical values do not cause errors during transformation.

### Train-Test Split

The dataset was divided into 80% training data and 20% testing data, resulting in 3,302 training observations and 826 testing observations.

Stratified sampling was used to preserve the original class distribution of the target variable. The proportions of both classes remain nearly identical in the training and testing sets, which helps provide a representative evaluation of the trained models.

## Model Training and Registration with Experimentation Tracking

Multiple classification algorithms will be trained and evaluated to identify the best-performing model for predicting whether a customer will purchase the Wellness Tourism Package.

The models will be compared using Accuracy, Precision, Recall, F1-score, and ROC-AUC. ROC-AUC and F1-score will receive particular attention because the target variable is imbalanced.

````python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    BaggingClassifier,
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier
)
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

# Define baseline models
models = {
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Bagging": BaggingClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(random_state=42),
    "AdaBoost": AdaBoostClassifier(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42)
}

results = []

# Train and evaluate each model
for name, model in models.items():
    model.fit(X_train_processed, y_train)

    y_pred = model.predict(X_test_processed)
    y_prob = model.predict_proba(X_test_processed)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

baseline_results = pd.DataFrame(results)

baseline_results.sort_values(
    by="ROC-AUC",
    ascending=False
).reset_index(drop=True)
````

````python
from sklearn.model_selection import GridSearchCV

# Hyperparameter grid for Random Forest
rf_param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [None, 10, 20],
    "min_samples_split": [2, 5],
    "min_samples_leaf": [1, 2]
}

rf_grid = GridSearchCV(
    estimator=RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ),
    param_grid=rf_param_grid,
    scoring="roc_auc",
    cv=3,
    n_jobs=-1,
    verbose=1
)

rf_grid.fit(X_train_processed, y_train)

print("Best Random Forest Parameters:")
print(rf_grid.best_params_)

print("\nBest Cross-Validation ROC-AUC:")
print(rf_grid.best_score_)
````

### Random Forest Hyperparameter Tuning

Random Forest was tuned using GridSearchCV with 3-fold cross-validation and ROC-AUC as the optimization metric.

A total of 24 hyperparameter combinations were evaluated. The best configuration uses 200 estimators, no maximum depth restriction, a minimum sample split of 2, and a minimum sample leaf size of 1.

The best cross-validation ROC-AUC obtained during tuning was 0.9351.

````python
# Get the best tuned Random Forest model
best_rf = rf_grid.best_estimator_

# Predictions on the test set
y_pred_rf = best_rf.predict(X_test_processed)
y_prob_rf = best_rf.predict_proba(X_test_processed)[:, 1]

# Evaluate tuned Random Forest
tuned_rf_results = {
    "Model": "Tuned Random Forest",
    "Accuracy": accuracy_score(y_test, y_pred_rf),
    "Precision": precision_score(y_test, y_pred_rf, zero_division=0),
    "Recall": recall_score(y_test, y_pred_rf, zero_division=0),
    "F1 Score": f1_score(y_test, y_pred_rf, zero_division=0),
    "ROC-AUC": roc_auc_score(y_test, y_prob_rf)
}

print("Tuned Random Forest Test Results:")
for metric, value in tuned_rf_results.items():
    if metric != "Model":
        print(f"{metric}: {value:.4f}")
````

### Tuned Random Forest Evaluation

The tuned Random Forest improves upon the baseline Random Forest across all evaluated metrics.

The ROC-AUC increased from 0.9660 to 0.9692, while the F1-score increased from 0.7154 to 0.7231. Precision and recall also improved slightly.

These results indicate that the selected hyperparameters provide a better-performing Random Forest model on the unseen test data.

````python
# Hyperparameter grid for Bagging
bagging_param_grid = {
    "n_estimators": [50, 100, 200],
    "max_samples": [0.5, 0.75, 1.0],
    "max_features": [0.5, 0.75, 1.0]
}

bagging_grid = GridSearchCV(
    estimator=BaggingClassifier(
        random_state=42,
        n_jobs=-1
    ),
    param_grid=bagging_param_grid,
    scoring="roc_auc",
    cv=3,
    n_jobs=-1,
    verbose=1
)

bagging_grid.fit(X_train_processed, y_train)

print("Best Bagging Parameters:")
print(bagging_grid.best_params_)

print("\nBest Cross-Validation ROC-AUC:")
print(bagging_grid.best_score_)
````

### Bagging Hyperparameter Tuning

Bagging was tuned using GridSearchCV with 3-fold cross-validation and ROC-AUC as the optimization metric.

A total of 27 hyperparameter combinations were evaluated. The best configuration uses 200 estimators, 100% of the training samples for each estimator, and 75% of the available features.

The best cross-validation ROC-AUC obtained was 0.9402.

````python
# Get the best tuned Bagging model
best_bagging = bagging_grid.best_estimator_

# Predictions on the test set
y_pred_bagging = best_bagging.predict(X_test_processed)
y_prob_bagging = best_bagging.predict_proba(X_test_processed)[:, 1]

# Evaluate tuned Bagging
tuned_bagging_results = {
    "Model": "Tuned Bagging",
    "Accuracy": accuracy_score(y_test, y_pred_bagging),
    "Precision": precision_score(y_test, y_pred_bagging, zero_division=0),
    "Recall": recall_score(y_test, y_pred_bagging, zero_division=0),
    "F1 Score": f1_score(y_test, y_pred_bagging, zero_division=0),
    "ROC-AUC": roc_auc_score(y_test, y_prob_bagging)
}

print("Tuned Bagging Test Results:")
for metric, value in tuned_bagging_results.items():
    if metric != "Model":
        print(f"{metric}: {value:.4f}")
````

### Tuned Bagging Evaluation

The tuned Bagging model outperforms the tuned Random Forest model across all evaluation metrics.

It achieves an accuracy of 0.9237, precision of 0.9364, recall of 0.6478, F1-score of 0.7658, and ROC-AUC of 0.9740.

The improvement in recall and F1-score is particularly important because the dataset contains an imbalanced target variable. Based on the current experiments, Tuned Bagging is the strongest-performing model.

### Gradient Boosting Hyperparameter Tuning

Gradient Boosting was tuned using GridSearchCV with 3-fold cross-validation and ROC-AUC as the optimization metric.

## Streamlit App

Please ensure that the web app script is named `app.py`.

````python
%%writefile ../deployment/app.py

import streamlit as st
import pandas as pd
import joblib
from huggingface_hub import hf_hub_download

# -----------------------------
# Load model from Hugging Face
# -----------------------------
MODEL_REPO = "rhythm0511/tourism-purchase-prediction"

model_path = hf_hub_download(
    repo_id=MODEL_REPO,
    filename="final_model.joblib",
    repo_type="model"
)

model = joblib.load(model_path)

# -----------------------------
# Streamlit UI
# -----------------------------
st.set_page_config(
    page_title="Wellness Tourism Package Prediction",
    page_icon="🏖️"
)

st.title("🏖️ Wellness Tourism Package Prediction")
st.write("Predict whether a customer is likely to purchase the package.")

st.subheader("Customer Information")

age = st.number_input("Age", min_value=18, max_value=100, value=30)
type_of_contact = st.selectbox(
    "Type of Contact",
    ["Self Enquiry", "Company Invited"]
)
city_tier = st.selectbox("City Tier", [1, 2, 3])
occupation = st.selectbox(
    "Occupation",
    ["Salaried", "Free Lancer", "Small Business", "Large Business"]
)
gender = st.selectbox("Gender", ["Male", "Female"])

number_of_person_visiting = st.number_input(
    "Number of Persons Visiting", min_value=1, value=2
)

preferred_property_star = st.selectbox(
    "Preferred Property Star", [3, 4, 5]
)

marital_status = st.selectbox(
    "Marital Status",
    ["Single", "Divorced", "Married", "Unmarried"]
)

number_of_trips = st.number_input(
    "Number of Trips", min_value=0, value=2
)

passport = st.selectbox("Passport", [0, 1])
own_car = st.selectbox("Own Car", [0, 1])

number_of_children_visiting = st.number_input(
    "Number of Children Visiting", min_value=0, value=0
)

designation = st.selectbox(
    "Designation",
    ["Manager", "Executive", "Senior Manager", "AVP", "VP"]
)

monthly_income = st.number_input(
    "Monthly Income", min_value=0.0, value=20000.0
)

pitch_satisfaction_score = st.selectbox(
    "Pitch Satisfaction Score", [1, 2, 3, 4, 5]
)

product_pitched = st.selectbox(
    "Product Pitched",
    ["Deluxe", "Basic", "Standard", "Super Deluxe", "King"]
)

number_of_followups = st.number_input(
    "Number of Followups", min_value=0, value=3
)

duration_of_pitch = st.number_input(
    "Duration of Pitch", min_value=0, value=15
)

if st.button("Predict Purchase"):
    
    input_data = pd.DataFrame([{
        "Age": age,
        "TypeofContact": type_of_contact,
        "CityTier": city_tier,
        "Occupation": occupation,
        "Gender": gender,
        "NumberOfPersonVisiting": number_of_person_visiting,
        "PreferredPropertyStar": preferred_property_star,
        "MaritalStatus": marital_status,
        "NumberOfTrips": number_of_trips,
        "Passport": passport,
        "OwnCar": own_car,
        "NumberOfChildrenVisiting": number_of_children_visiting,
        "Designation": designation,
        "MonthlyIncome": monthly_income,
        "PitchSatisfactionScore": pitch_satisfaction_score,
        "ProductPitched": product_pitched,
        "NumberOfFollowups": number_of_followups,
        "DurationOfPitch": duration_of_pitch
    }])

    prediction = model.predict(input_data)[0]
    probability = model.predict_proba(input_data)[0][1]

    if prediction == 1:
        st.success(
            f"Customer is likely to purchase the package "
            f"(Probability: {probability:.2%})"
        )
    else:
        st.warning(
            f"Customer is unlikely to purchase the package "
            f"(Probability: {probability:.2%})"
        )
````

# MLOps Pipeline with Github Actions Workflow

**Note:**

1. Before running the file below, make sure to add the HF_TOKEN to your GitHub secrets to enable authentication between GitHub and Hugging Face.
2. The below code is for a sample YAML file that can be updated as required to meet the requirements of this project.

```
name: Tourism Project Pipeline

on:
  push:
    branches:
      - main  # Automatically triggers on push to the main branch

jobs:

  register-dataset:
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v3
      - name: Install Dependencies
        run: <add_code_here>
      - name: Upload Dataset to Hugging Face Hub
        env:
          HF_TOKEN: ${{ secrets.HF_TOKEN }}
        run: <add_code_here>

  data-prep:
    needs: register-dataset
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v3
      - name: Install Dependencies
        run: <add_code_here>
      - name: Run Data Preparation
        env:
          HF_TOKEN: ${{ secrets.HF_TOKEN }}
        run: <add_code_here>


  model-traning:
    needs: data-prep
    runs-on: ubuntu-latest
    steps:
      - uses: actions/checkout@v3
      - name: Install Dependencies
        run: <add_code_here>
      - name: Start MLflow Server
        run: |
          nohup mlflow ui --host 0.0.0.0 --port 5000 &  # Run MLflow UI in the background
          sleep 5  # Wait for a moment to let the server starts
      - name: Model Building
        env:
          HF_TOKEN: ${{ secrets.HF_TOKEN }}
        run: <add_code_here>


  deploy-hosting:
    runs-on: ubuntu-latest
    needs: [model-traning,data-prep,register-dataset]
    steps:
      - uses: actions/checkout@v3
      - name: Install Dependencies
        run: <add_code_here>
      - name: Push files to Frontend Hugging Face Space
        env:
          HF_TOKEN: ${{ secrets.HF_TOKEN }}
        run: <add_code_here>

```

**Note:** To use this YAML file for our use case, we need to

1. Go to the GitHub repository for the project
2. Create a folder named ***.github/workflows/***
3. In the above folder, create a file named ***pipeline.yml***
4. Copy and paste the above content for the YAML file into the ***pipeline.yml*** file
